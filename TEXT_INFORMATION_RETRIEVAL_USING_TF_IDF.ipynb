```python
df = pd.DataFrame({
    "Document_ID": [f"D{i+1}" for i in range(len(documents))],
    "Text": documents
})

df
```

```python
def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    words = text.split()
    words = [word for word in words if word not in stop_words]
    words = [lemmatizer.lemmatize(word) for word in words]
    return words
```

```python
df["Processed_Text"] = df["Text"].apply(preprocess)

df[["Document_ID", "Processed_Text"]]
```

```python
vocabulary = sorted(
    set(word for doc in df["Processed_Text"] for word in doc)
)

print("Vocabulary Size:", len(vocabulary))
```

```python
def calculate_tf(document):
    tf = {}
    total_words = len(document)

    for word in vocabulary:
        tf[word] = document.count(word) / total_words

    return tf
```

```python
tf_matrix = []

for doc in df["Processed_Text"]:
    tf_matrix.append(calculate_tf(doc))

tf_df = pd.DataFrame(tf_matrix, index=df["Document_ID"])

tf_df.round(4)
```

```python
N = len(df)

idf = {}

for word in vocabulary:
    df_count = sum(1 for doc in df["Processed_Text"] if word in doc)
    idf[word] = math.log((N + 1) / (df_count + 1)) + 1

idf_df = pd.DataFrame(
    list(idf.items()),
    columns=["Term", "IDF"]
)

idf_df
```

```python
tfidf_matrix = []

for doc in df["Processed_Text"]:
    tf = calculate_tf(doc)

    tfidf = {}

    for word in vocabulary:
        tfidf[word] = tf[word] * idf[word]

    tfidf_matrix.append(tfidf)

tfidf_df = pd.DataFrame(
    tfidf_matrix,
    index=df["Document_ID"]
)

tfidf_df.round(4)
```

```python
query = input("Enter your query: ")
```

```python
processed_query = preprocess(query)

query_terms = [word for word in processed_query if word in vocabulary]

print(query_terms)
```

```python
scores = {}

for doc in tfidf_df.index:
    score = 0

    for term in query_terms:
        score += tfidf_df.loc[doc, term]

    scores[doc] = score
```

```python
results = pd.DataFrame(
    list(scores.items()),
    columns=["Document_ID", "Score"]
)

results = results.sort_values(
    by="Score",
    ascending=False
)

print(results)
```

```python
top_doc = results.iloc[0]["Document_ID"]

print("Most Relevant Document ID:", top_doc)
```
